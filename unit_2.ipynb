```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier(["I like learning",
            "i hate this so much"])
```

```python
from transformers import AutoTokenizer
checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

```python
raw_inputs = ["i love this course"]

inputs = tokenizer(raw_inputs, padding=True, truncation=True, return_tensors='pt')
```

```python
inputs
```

```python
from transformers import AutoModel

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModel.from_pretrained(checkpoint)
```

```python
outputs = model(**inputs)
```

```python
print(outputs.last_hidden_state.shape)
```

```python
from transformers import AutoModelForSequenceClassification

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
outputs = model(**inputs)
```

```python
print(outputs.logits.shape)
```

```python
print(outputs.logits)
```

```python
import torch

predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)
print(predictions)
```

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained("bert-base-cased")
```

```python
tokenizer("Using a Transformer network is simple")
```

```python
tokens = tokenizer.tokenize("Let's try tokenizer")
print(tokens)
```

```python
input_ids = tokenizer.convert_tokens_to_ids(tokens)
print(input_ids)
```

```python
final_inputs = tokenizer.prepare_for_model(tokens)
print(final_inputs)
input_ids = tokenizer.convert_tokens_to_ids(final_inputs["input_ids"])
print(input_ids)
```

```python
tokenizer.decode(input_ids)
```
